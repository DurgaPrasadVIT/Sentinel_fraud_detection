```python
import pandas as pd

train = pd.read_parquet("../data/features/train.parquet")

cols = ["amount", "count_1h", "count_24h", "sum_24h",
        "amount_vs_avg", "secs_since_last", "is_night", "distance_km"]

train.groupby("is_fraud")[cols].median().T.round(2)
```

## Feature check (medians, normal vs fraud)
- Strong: amount (8x), sum_24h (10x), amount_vs_avg (0.66 vs 5.23), secs_since_last (4.6h vs 1.4h), is_night
- Weak at median: count_1h, count_24h (may still help in extreme cases)
- Useless here: distance_km (no difference in this synthetic data)
- Idea: "impossible travel" = distance from previous payment / time between them
